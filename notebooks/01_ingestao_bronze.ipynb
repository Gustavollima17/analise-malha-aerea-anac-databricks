```python
import re
from pyspark.sql import functions as F

caminho_arquivos = "/Volumes/workspace/default/dados_anac/CSVs/VRA_*.csv"

df_raw = (spark.read
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "UTF-8")
    .option("skipRows", 1)
    .option("inferSchema", "true")
    .csv(caminho_arquivos)
)

# Limpa espaços e caracteres especiais dos nomes das colunas para compatibilidade com Delta Lake
colunas_limpas = [re.sub(r'[^a-zA-Z0-9_]', '_', col.strip()) for col in df_raw.columns]
df_limpo = df_raw.toDF(*colunas_limpas)

df_bronze = df_limpo.withColumn("data_ingestao", F.current_timestamp())

df_bronze.write.format("delta").mode("overwrite").saveAsTable("bronze_voos_anac")

print(f"Total de voos carregados na camada Bronze: {df_bronze.count()}")
df_bronze.printSchema()
display(df_bronze.limit(10))
```
